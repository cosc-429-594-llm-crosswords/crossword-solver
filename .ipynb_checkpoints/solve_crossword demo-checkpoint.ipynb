```python
from src.classes.clue import Clue
from src.classes.crossword_puzzle import CrosswordPuzzle
from src.classes.guesses import Guess
from src.constants import CLUE_ID, PUZ_FILE_DIR
```

```python
crossword_puzzle = CrosswordPuzzle(f"{PUZ_FILE_DIR}/nytm_2025_01_01.puz")
guesses: dict[CLUE_ID, list[Guess]] = {}
```

```python
clues = [Clue(text='Month that was the first of the new year in early 3-Down calendars', length=5, number=1, direction='across', row=0, col=0),
 Clue(text='Not-so-nice magic spells', length=5, number=5, direction='down', row=0, col=4),
 Clue(text='Maker of Ironman Triathlon watches', length=5, number=7, direction='across', row=2, col=0),
 Clue(text="Francophile's farewell", length=5, number=2, direction='down', row=0, col=1),
 Clue(text="Feature of a cockatoo's head", length=5, number=4, direction='down', row=0, col=3),
 Clue(text='Like Janus, the god of beginnings', length=5, number=3, direction='down', row=0, col=2),
 Clue(text='Searches (for)', length=5, number=9, direction='across', row=4, col=0),
 Clue(text='Soccer contest', length=5, number=1, direction='down', row=0, col=0),
 Clue(text='Love so much', length=5, number=6, direction='across', row=1, col=0),
 Clue(text='Stop', length=5, number=8, direction='across', row=3, col=0)]
```

```python
first_time_calling_5_down = True

def mock_get_guesses_for_clue_using_llm(clue: Clue) -> list[Guess]:
    clue_key = (clue.number, clue.direction)

    if clue_key == (1, "across"):
        return [
            Guess(answer="JANUS", confidence_score=95, explanation="The god of beginnings, though not the month itself."),
            Guess(answer="MARCH", confidence_score=60, explanation="The first month in the early Roman calendar."),
            Guess(answer="APRIL", confidence_score=40, explanation="Associated with spring, but historically incorrect for 'first'."),
            Guess(answer="MARDS", confidence_score=20, explanation="A phonetic misspelling of March.")
        ]
    elif clue_key == (7, "across"):
        return [
            Guess(answer="TIMEX", confidence_score=99, explanation="The official maker of Ironman watches."),
            Guess(answer="CASIO", confidence_score=45, explanation="Common digital watch brand, but not the Ironman maker."),
            Guess(answer="SEIKO", confidence_score=30, explanation="Well-known watch brand, wrong niche."),
            Guess(answer="GARMI", confidence_score=25, explanation="Truncated 'Garmin', who makes Ironman-compatible GPS watches.")
        ]
    elif clue_key == (6, "across"):
        return [
            Guess(answer="ADORE", confidence_score=95, explanation="Standard 5-letter synonym for intense love."),
            Guess(answer="LOVES", confidence_score=80, explanation="Simple, direct synonym."),
            Guess(answer="PRIZE", confidence_score=50, explanation="To value something greatly."),
            Guess(answer="EXALT", confidence_score=40, explanation="To hold in high regard, slightly different nuance.")
        ]
    elif clue_key == (8, "across"):
        return [
            Guess(answer="CEASE", confidence_score=90, explanation="Formal synonym for stop."),
            Guess(answer="HALTS", confidence_score=88, explanation="Common crossword filler for stop."),
            Guess(answer="ABORT", confidence_score=60, explanation="To stop a process abruptly."),
            Guess(answer="STAND", confidence_score=30, explanation="As in 'come to a stand-still'.")
        ]
    elif clue_key == (9, "across"):
        return [
            Guess(answer="HUNTS", confidence_score=95, explanation="Matches 'searches (for)' perfectly."),
            Guess(answer="SEEKS", confidence_score=92, explanation="Strong synonym for searching."),
            Guess(answer="LOOKS", confidence_score=75, explanation="Simple synonym, usually requires 'for'."),
            Guess(answer="SCOUR", confidence_score=65, explanation="To search thoroughly."),
            Guess(answer="PORES", confidence_score=40, explanation="To search through something, usually 'pores over'.")
        ]
    elif clue_key == (1, "down"):
        return [
            Guess(answer="MATCH", confidence_score=98, explanation="A standard term for a soccer contest."),
            Guess(answer="DERBY", confidence_score=70, explanation="A specific type of soccer match between local rivals."),
            Guess(answer="GAMES", confidence_score=55, explanation="General term, though usually plural."),
            Guess(answer="FINAL", confidence_score=40, explanation="A specific type of high-stakes contest.")
        ]
    elif clue_key == (2, "down"):
        return [
            Guess(answer="ADIEU", confidence_score=99, explanation="The quintessential Francophile farewell."),
            Guess(answer="SALUT", confidence_score=60, explanation="Can mean both hello and goodbye in French."),
            Guess(answer="CONGE", confidence_score=45, explanation="Formal leave-taking."),
            Guess(answer="AVANT", confidence_score=15, explanation="French word, but means 'before', not goodbye.")
        ]
    elif clue_key == (3, "down"):
        return [
            Guess(answer="ROMAN", confidence_score=95, explanation="Janus is a Roman god."),
            Guess(answer="FACED", confidence_score=75, explanation="Janus is famously 'two-faced'."),
            Guess(answer="DUALS", confidence_score=50, explanation="Referring to Janus's dual nature."),
            Guess(answer="LATIN", confidence_score=40, explanation="Related to the culture, but less specific to the god.")
        ]
    elif clue_key == (4, "down"):
        return [
            Guess(answer="CREST", confidence_score=99, explanation="The feathers on a cockatoo's head."),
            Guess(answer="PLUME", confidence_score=70, explanation="A fancy word for a feather."),
            Guess(answer="TUFTS", confidence_score=50, explanation="Small bunches of feathers."),
            Guess(answer="PEAKS", confidence_score=30, explanation="Metaphorical description of the head shape.")
        ]
    elif clue_key == (5, "down"):
        global first_time_calling_5_down
        if first_time_calling_5_down:
            first_time_calling_5_down = False
            return [
                Guess(answer="CURSE", confidence_score=95, explanation="Direct synonym for a bad spell."),
            ]
        else:
            return [
                Guess(answer="PPPPP", confidence_score=95, explanation="Direct synonym for a bad spell."),
                Guess(answer="HEXES", confidence_score=90, explanation="Classic 5-letter word for bad spells."),
            ]

    return []
```

```python
clue_index = 0

while not crossword_puzzle.is_solved:
    clue = clues[clue_index]
    print(f"Attempting to solve clue {clue.number} {clue.direction} - {clue.text}")

    if clue.id not in guesses:
        print(f"Mock generating guesses for clue: {clue.number} {clue.direction} - {clue.text}")
        clue_guesses = mock_get_guesses_for_clue_using_llm(clue)
        guesses[clue.id] = clue_guesses

    clue_guesses = guesses[clue.id]

    if len(clue_guesses) == 0:
        print(f"No guesses left for clue: {clue.number} {clue.direction} - {clue.text}, backtracking...")
        guesses.pop(clue.id)

        clue_index -= 1
        crossword_puzzle.remove_answer(clues[clue_index])
        continue

    best_guess = max(clue_guesses, key=lambda g: g.confidence_score)

    try:
        crossword_puzzle.set_answer(clue, best_guess.answer)
        clue_guesses.remove(best_guess)

        print(f"Set answer for clue {clue.number} {clue.direction} to '{best_guess.answer}'")
        clue_index += 1
    except Exception as e:

        clue_guesses.remove(best_guess)
        print(f"Error setting answer for clue {clue.number} {clue.direction}: {e}")

crossword_puzzle.print_grid()
```
