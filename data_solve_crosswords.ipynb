This notebook solves a batch of crosswords (defined in monday_batches.csv) with a parameter-set ordering method. The clue guess generation will use the best method found by data gathering (assume that it will be self-consistency + suggestions). 

PARAMETERS:
- RANKING: 0 = none, 1 = vagueness/complexity, 2 = vagueness/complexity + ordering
- BATCH: The batch number to solve (e.g., batch_no=1 will solve the first batch in monday_batches.csv)

```python
# import libraries
import argparse
import pandas as pd
from datetime import date

from solve_crossword import solve_crossword
from src.constants import RANKING_ALGORITHMS
```

```python
# Parameters
RANKING: RANKING_ALGORITHMS = "VAGUENESS_AND_COMPLEXITY_PLUS_KNOWN_LETTERS"
BATCH_NUM = 2 # out of 10, because total 50 puzzles

# Constants
BATCH_SIZE = 5 
start_date = date(2025, 1, 1)
end_date   = date(2026, 1, 1)
```

```python
def get_files_in_batch(start: date, end: date, batch_num: int, batch_size: int = 5,) -> list[str]:
    """Return all file names in batch of Mondays."""
    mondays = pd.date_range(start=start_date, end=end_date, freq='W-MON').tolist()

    batch_list = []
    for monday in mondays[(batch_num-1) * batch_size: batch_num * batch_size]:
        batch_list.append(monday.strftime('nytm_%Y_%m_%d'))

    return batch_list
```

```python
parser = argparse.ArgumentParser()
parser.add_argument("--ranking-alg", type=int, required=True)
parser.add_argument("--batch-num", type=int, required=True)

batch_list = get_files_in_batch(start_date, end_date, batch_num=BATCH_NUM, batch_size=BATCH_SIZE)

for file in batch_list: 
    solve_crossword(file, RANKING)
```
