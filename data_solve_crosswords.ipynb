This notebook solves a batch of crosswords (defined in monday_batches.csv) with a parameter-set ordering method. The clue guess generation will use the best method found by data gathering (assume that it will be self-consistency + suggestions). 

PARAMETERS:
- RANKING: 0 = none, 1 = vagueness/complexity, 2 = vagueness/complexity + ordering
- BATCH: The batch number to solve (e.g., batch_no=1 will solve the first batch in monday_batches.csv)

```python
# Parameters
RANKING = 2
BATCH = 2

# Constants
BATCH_SIZE = 5 # will need to 
```

```python
# import libraries
import pandas as pd
from datetime import date, timedelta
import sys
```

```python
def get_mondays_in_range(start: date, end: date) -> list[date]:
    """Return all Mondays between start and end (inclusive)."""
    # Advance start to the first Monday on or after it
    days_until_monday = (7 - start.weekday()) % 7  # weekday(): Mon=0, Sun=6
    first_monday = start + timedelta(days=days_until_monday)

    mondays = []
    current = first_monday
    while current <= end:
        mondays.append(current)
        current += timedelta(weeks=1)

    return mondays


def batch_mondays(start: date, end: date, batch_size: int = 5):
    """Yield batches of `batch_size` Mondays within the date range."""
    mondays = get_mondays_in_range(start, end)
    for i in range(0, len(mondays), batch_size):
        yield mondays[i : i + batch_size]
```

```python
start_date = date(2025, 1, 1)
end_date   = date(2026, 1, 1)

batches = list(batch_mondays(start_date, end_date, batch_size=BATCH_SIZE))

if BATCH < 1 or BATCH > len(batches):
    print(f"Error: batch {BATCH} out of range (1–{len(batches)})")

batch = batches[BATCH - 1]
print(f"Batch {BATCH}:")

batch_list = []
for d in batch:
    batch_list.append(d.strftime('nytm_%Y_%m_%d.puz'))

batch_list
```
