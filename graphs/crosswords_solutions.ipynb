# New York Times Mini Crosswords Monday Puzzle Solutions
AI was used for formatting each solution in the file

```python
import os
from datetime import date, timedelta
import puz

PUZ_FOLDER = "../puz_files"
OUTPUT_FILE = "monday_solutions.txt"

start = date(2025, 1, 6)
end = date(2025, 12, 31)
```

```python
all_files = os.listdir(PUZ_FOLDER)

with open(OUTPUT_FILE, "w", encoding="utf-8") as out:
    current = start

    while current <= end:
        current_str = current.strftime("%Y_%m_%d")

        matching_file = None

        for file in all_files:
            if current_str in file and file.endswith(".puz"):
                matching_file = file
                break

        file_path = os.path.join(PUZ_FOLDER, matching_file)

        puzzle = puz.read(file_path)
        
        width = puzzle.width
        solution = puzzle.solution

        formatted_solution = ""

        for i in range(0, len(solution), width):
            row = solution[i:i+width]
            formatted_solution += " ".join(row) + "\n"

        out.write(f"File: {matching_file}\n")
        out.write(f"Date: {current}\n")
        out.write("Solution:\n")
        out.write(formatted_solution)
        out.write("\n")
        out.write("=" * 40 + "\n\n")

        current += timedelta(days=7)
```
