```python
import pandas as pd

from src.classes.crossword_puzzle import CrosswordPuzzle
from src.constants import PUZ_FILE_DIR
```

```python
import pandas as pd

outlet = "nytm"
start_date = '2015-01-01'
end_date = '2020-12-31'
date_range = pd.date_range(start=start_date, end=end_date, freq='D')

all_puzzles = []

for date in date_range:
    year, month, day = date.year, date.month, date.day
    day_of_week = date.day_name()

    file = f"{PUZ_FILE_DIR}/{outlet}_{year}_{month:02d}_{day:02d}.puz"

    try:
        puzzle = CrosswordPuzzle(file)
        clues = puzzle.get_clues()

        temp_df = pd.DataFrame({
            "file": [file] * len(clues),
            "date": [date.strftime('%Y-%m-%d')] * len(clues),
            "day_of_week": [day_of_week] * len(clues), # Added column
            "number": [clue.number for clue in clues],
            "direction": [clue.direction for clue in clues],
            "length": [clue.length for clue in clues],
            "text": [clue.text for clue in clues],
            "solution": [puzzle.get_solution(clue) for clue in clues]
        })

        all_puzzles.append(temp_df)
    except Exception as e:
        print(f"Error occurred while processing {file}: {e}")
        continue

final_df = pd.concat(all_puzzles, ignore_index=True)
```
